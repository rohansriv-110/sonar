## 5. Visualize loss vs. predicted probability

This is the classic "why cross-entropy punishes confident wrong answers so harshly" plot.


```
import matplotlib.pyplot as plt

p = np.linspace(0.001, 0.999, 200)

loss_if_true_1 = -np.log(p)        # loss when true label is 1
loss_if_true_0 = -np.log(1 - p)    # loss when true label is 0

plt.figure(figsize=(7,5))
plt.plot(p, loss_if_true_1, label="y_true = 1")
plt.plot(p, loss_if_true_0, label="y_true = 0")
plt.xlabel("predicted probability p")
plt.ylabel("loss")
plt.title("Binary cross-entropy: loss vs predicted probability")
plt.legend()
plt.grid(True)
plt.show()

# Question: at p=0.5 (total uncertainty), what's the loss for both curves? Why do they meet there?
```
